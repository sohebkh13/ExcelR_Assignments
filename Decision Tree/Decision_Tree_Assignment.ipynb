Step 1: Data Preparation
Start by importing the necessary libraries and loading the dataset.

```python
import pandas as pd
data=pd.read_excel('heart_disease.xlsx', sheet_name='Heart_disease')
data.head()
```

Step 2: Exploratory Data Analysis (EDA)
Explore the data to understand its structure, detect missing values, and visualize distributions.

```python
# Task 2: Exploratory Data Analysis (EDA)
# Check for missing values in each column to identify potential data issues
print("Missing values in each column:")
print(data.isnull().sum())
```

```python
# Display summary statistics to understand the distribution and spread of numerical features
print("\nSummary statistics:")
data.describe()
```

```python
# Visualize the distribution of target variable 'Heart disease'
# Import visualization libraries for plotting
import matplotlib.pyplot as plt
import seaborn as sns

# Visualize the distribution of numerical features using histograms
data.hist(bins=15, figsize=(15, 10))
plt.suptitle("Histograms of Numerical Features")
plt.show()
```

```python
# Use box plots to detect outliers in numerical features
plt.figure(figsize=(15, 10))
sns.boxplot(data=data)
plt.title("Box Plots of Numerical Features")
plt.xticks(rotation=90)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Assuming `data` is your DataFrame

# Apply one-hot encoding to non-numeric columns (convert categorical data to numeric)
data_encoded = pd.get_dummies(data, drop_first=True)

# Create correlation matrix including the numeric and one-hot encoded columns
plt.figure(figsize=(12, 8))
sns.heatmap(data_encoded.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix with One-Hot Encoding")
plt.show()
```
