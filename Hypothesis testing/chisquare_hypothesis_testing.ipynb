```python
# Chi-Square Test for Independence

import numpy as np
from scipy import stats

# Given Data (Observed Frequencies)
observed = np.array([
    [50, 70],   # Very Satisfied
    [80, 100],  # Satisfied
    [60, 90],   # Neutral
    [30, 50],   # Unsatisfied
    [20, 50]    # Very Unsatisfied
])

# Compute Row and Column Totals
row_totals = np.sum(observed, axis=1).reshape(-1, 1)
col_totals = np.sum(observed, axis=0)
grand_total = np.sum(observed)

# Compute Expected Frequencies
expected = (row_totals @ col_totals.reshape(1, -1)) / grand_total

# Compute Chi-Square Statistic
chi_square_stat = np.sum(((observed - expected) ** 2) / expected)

# Degrees of Freedom (df)
df = (observed.shape[0] - 1) * (observed.shape[1] - 1)

# Critical Value (alpha = 0.05)
alpha = 0.05
critical_value = stats.chi2.ppf(1 - alpha, df)

# Decision
if chi_square_stat > critical_value:
    decision = "Reject the null hypothesis: There is an association between device type and satisfaction."
else:
    decision = "Fail to reject the null hypothesis: No significant association found."

# Print Results
print("Chi-Square Statistic:", round(chi_square_stat, 4))
print("Degrees of Freedom:", df)
print("Critical Value:", round(critical_value, 4))
print("Decision:", decision)
```

```python
# Hypothesis Testing for Weekly Operating Costs

import numpy as np
from scipy import stats

# Given Data
sample_mean = 3050  # Sample Mean Weekly Cost
theoretical_mean = 1000 + (5 * 600)  # Theoretical Mean W = 1000 + 5*X (X=600)
population_std = 5 * 25  # Population Standard Deviation
sample_size = 25  # Sample Size

# Compute Test Statistic (z-score)
test_stat = (sample_mean - theoretical_mean) / (population_std / np.sqrt(sample_size))

# Critical Value for One-Tailed Test (alpha = 0.05)
alpha = 0.05
critical_value = stats.norm.ppf(1 - alpha)

# Decision
if test_stat > critical_value:
    decision = "Reject the null hypothesis: The weekly operating cost has increased."
else:
    decision = "Fail to reject the null hypothesis: No strong evidence of an increase."

# Print Results
print("Test Statistic (z-score):", round(test_stat, 4))
print("Critical Value (alpha = 0.05):", round(critical_value, 4))
print("Decision:", decision)
```
