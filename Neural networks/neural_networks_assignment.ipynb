In this assignment, you will be tasked with developing a classification model using Artificial Neural Networks (ANNs) to classify data points from the "Alphabets_data.csv" dataset into predefined categories of alphabets.
```python
# Step 1: Import Required Libraries
# Import necessary libraries for data handling, visualization, and building the neural network.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import Adam
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
# Step 2: Load and Explore the Dataset
# Load the "Alphabets_data.csv" dataset.
dataset_path = 'Alphabets_data.csv'
data = pd.read_csv(dataset_path)

# Display the first few rows of the dataset
data.head()
```

```python
# Summarize dataset properties
print("Dataset Info:")
data.info()
print("\nDataset Summary:\n", data.describe())

# Check for class distribution
plt.figure(figsize=(8, 6))
sns.countplot(data=data, x=data.columns[-1])
plt.title("Class Distribution")
plt.xlabel("Alphabet Class")
plt.ylabel("Count")
plt.show()
```

```python
# Step 3: Data Preprocessing
# Handling missing values (if any)
print("\nMissing Values:\n", data.isnull().sum())
data.dropna(inplace=True)
```

```python
# Convert categorical data to numeric using LabelEncoder
label_encoder = LabelEncoder()
for column in data.select_dtypes(include=['object']).columns:
    data[column] = label_encoder.fit_transform(data[column])

# Separate features and target variable
X = data.iloc[:, :-1]
y = data.iloc[:, -1]

# Normalize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42, stratify=y)
```

```python
# Step 4: Build the Basic ANN Model
# Define a function to create the ANN model
from tensorflow.keras import Input

def create_model(neurons=64, activation="relu", learning_rate=0.01, **kwargs):
    model = Sequential()
    model.add(Input(shape=(X_train.shape[1],)))  # Define the input shape explicitly
    model.add(Dense(neurons, activation=activation))
    model.add(Dense(neurons, activation=activation))
    model.add(Dense(len(np.unique(y)), activation="softmax"))
    optimizer = Adam(learning_rate=learning_rate)
    model.compile(optimizer=optimizer, loss="sparse_categorical_crossentropy", metrics=["accuracy"])
    return model

```

```python
# Instantiate the model
model = create_model()

# Train the model on the training set
history = model.fit(X_train, y_train, epochs=50, batch_size=32, validation_split=0.2, verbose=1)
```

```python
# Step 5: Evaluate the Basic Model
# Evaluate the model on the test set
test_loss, test_accuracy = model.evaluate(X_test, y_test, verbose=0)
print("\nTest Accuracy (Basic Model):", test_accuracy)
```

```python
# Generate predictions and classification report
from sklearn.metrics import classification_report, confusion_matrix,accuracy_score
y_pred = np.argmax(model.predict(X_test), axis=1)
print("\nClassification Report (Basic Model):\n", classification_report(y_test, y_pred))
```

```python
# Confusion matrix
conf_matrix = confusion_matrix(y_test, y_pred)
sns.heatmap(conf_matrix, annot=True, cmap='Blues', fmt='d')
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
# Step 6: Hyperparameter Tuning
# Define the parameter grid for grid search
param_grid = {
    "model__neurons": [32, 64, 128],
    "model__activation": ["relu", "tanh"],
    "model__learning_rate": [0.001, 0.01, 0.1],
}


```

```python
# Wrap the Keras model using KerasClassifier
from scikeras.wrappers import KerasClassifier

# Wrap the model
model_wrapper = KerasClassifier(model=create_model, verbose=0, epochs=50, batch_size=32)
```

```python
# Perform grid search with cross-validation

from sklearn.model_selection import GridSearchCV

grid = GridSearchCV(estimator=model_wrapper, param_grid=param_grid, cv=3, scoring="accuracy", verbose=2)
grid_result = grid.fit(X_train, y_train)
```

```python
# Display the best parameters and corresponding accuracy
print("\nBest Parameters:", grid_result.best_params_)
print("\nBest Cross-Validation Accuracy:", grid_result.best_score_)
```

```python
# Step 7: Evaluate the Tuned Model
# Evaluate the best model on the test set
best_model = grid_result.best_estimator_
tuned_test_accuracy = best_model.score(X_test, y_test)
print("\nTest Accuracy (Tuned Model):", tuned_test_accuracy)
```

```python
# Generate predictions and classification report for the tuned model
y_pred_tuned = best_model.predict(X_test)
print("\nClassification Report (Tuned Model):\n", classification_report(y_test, y_pred_tuned))

# Confusion matrix for the tuned model
conf_matrix_tuned = confusion_matrix(y_test, y_pred_tuned)
sns.heatmap(conf_matrix_tuned, annot=True, cmap='Blues', fmt='d')
plt.title("Confusion Matrix (Tuned Model)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```
