```python
# Step 1: Import Required Libraries
# Import necessary libraries for data handling, preprocessing, and association rule mining.
# Import necessary libraries
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder
```

```python
# Step 1: Load Dataset
file_path = "processed_online_retail.csv"
df = pd.read_csv(file_path, encoding="ISO-8859-1")  # Handle special characters
df.head()
```

```python
# Step 2: Data Preprocessing
# Remove any unwanted characters from column names
df.columns = df.columns.str.strip().str.replace('"', '')

# Remove duplicate transactions
df_cleaned = df.drop_duplicates()

# Fill missing values with empty strings (since transactions are lists)
df_cleaned = df_cleaned.fillna('')

# Convert the dataset into a list of transactions
transactions = df_cleaned.iloc[:, 0].apply(lambda x: x.split(','))  # Split items
transactions_list = transactions.tolist()  # Convert to list of lists

# Step 3: Convert Transactions to Binary Format
te = TransactionEncoder()
te_ary = te.fit(transactions_list).transform(transactions_list)
df_encoded = pd.DataFrame(te_ary, columns=te.columns_)

# Step 4: Apply Apriori Algorithm
min_support = 0.02  # Minimum support threshold
frequent_itemsets = apriori(df_encoded, min_support=min_support, use_colnames=True)

# Step 5: Generate Association Rules
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.3)

# Display frequent itemsets
# Display frequent itemsets
print("Frequent Itemsets:")
print(frequent_itemsets)

# Display association rules
print("Association Rules:")
print(rules)
```
